IMPORT Library

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Load dataset

```python
df = pd.read_csv("student_habits_performance.csv")

print(df.head())
```

1. Korelasi antara Study Hours, Attendance, dan Exam Score

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x=df['study_hours_per_day'], y=df['exam_score'], hue=df['attendance_percentage'], palette='coolwarm')
plt.xlabel("Study Hours per Day")
plt.ylabel("Exam Score")
plt.title("Korelasi Study Hours, Attendance, dan Exam Score")
plt.show()

plt.figure(figsize=(6,5))
sns.heatmap(df[['study_hours_per_day', 'attendance_percentage', 'exam_score']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Heatmap Korelasi")
plt.show()
```

2. Pengaruh Diet Quality, Exercise Frequency, dan Internet Quality terhadap Exam Score

```python
features = ['diet_quality', 'exercise_frequency', 'internet_quality']
plt.figure(figsize=(12,6))
for i, feature in enumerate(features):
    plt.subplot(1, 3, i+1)
    sns.boxplot(x=df[feature], y=df['exam_score'])
    plt.xlabel(feature.replace('_', ' ').title())
    plt.ylabel("Exam Score")
    plt.title(f"Exam Score berdasarkan {feature.replace('_', ' ').title()}")
plt.tight_layout()
plt.show()
```

3. Distribusi Exam Score

```python
plt.figure(figsize=(8,6))
sns.histplot(df['exam_score'], bins=20, kde=True, color="blue")
plt.xlabel("Exam Score")
plt.ylabel("Frequency")
plt.title("Distribusi Exam Score")
plt.show()
```

4. Perbandingan Exam Score berdasarkan Gender

```python
plt.figure(figsize=(8,6))
sns.barplot(x=df['gender'], y=df['exam_score'], ci=None, palette="pastel")
plt.xlabel("Gender")
plt.ylabel("Average Exam Score")
plt.title("Perbandingan Exam Score Berdasarkan Gender")
plt.show()
```
